# Data Description
- Background: 美國德州 Austin Animal Center 的動物結局（Outcome）資料，用於分析動物的品種、毛色、年齡、結局類型、結局時間等相關特徵，常用於動物福利研究、收容所管理效能分析與動物行為預測模型建立。
- date received: 20250928
- Path to data file: /data/animal_shelter/austin/Austin_Animal_Center_Outcomes.csv
- Unit of observation: 每一筆資料代表「一隻動物於 Austin Animal Center 的結局紀錄（Outcome Record）」。
- Sample period: 依資料內容顯示涵蓋 2013 年至 2020 年（不同動物紀錄略有差異）。
- Known issues:
    - Outcome Subtype 欄位多為缺失值。
    - 日期皆為字串格式（DateTime、Date of Birth、MonthYear），需轉換為 datetime 才能分析。
    - Sex upon Outcome 與 Age upon Outcome 為字串格式（如 “Neutered Male” 或 “2 years”），使用前需解析。
    - Breed（品種）多為混種描述，可能需拆解或重新分類。
    - Color 為非結構化文字（如 “Brown Tabby/White”），需整理後才能用於模型。

- Definition for each variable:
    - Animal ID: 動物唯一識別碼
    - Name: 動物名稱（部分缺失）
    - DateTime: 動物結局產生的日期與時間
    - MonthYear: 字串格式的年月資料（重複資訊，來源於 DateTime）
    - Date of Birth: 出生日期（部份缺失）
    - Outcome Type: 動物結局類型（如 Adoption、Transfer、Euthanasia、Rto-Adopt 等）
    - Outcome Subtype: 結局子分類（多為空值）
    - Animal Type: 動物種類（Dog、Cat、Other 等）
    - Sex upon Outcome: 結局時的性別與結紮狀態（如 Neutered Male、Intact Female）
    - Age upon Outcome: 動物結局時的年齡（字串格式：如 “2 years”、“4 months”）
    - Breed: 品種（含混種，如 Labrador Retriever Mix 等）
    - Color: 毛色描述（如 Brown Tabby/White）
