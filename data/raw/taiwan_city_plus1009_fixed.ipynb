# Data Description
- Background: 台灣寵物資料（加強版）。此資料包含更細部之寵物收容流動資訊，用於分析台灣各縣市對寵物之偏好與收容流程狀態，例如：期初收容量、移入來源、各類移出方式、期末收容量、滿載率等。
- date received: 20251003
- Path to data file: /data/pet_management/taiwan/monthly_city_reports/taiwan_city_plus1009.json
- Unit of observation: 每一筆資料代表「某一縣市在某一年月份的寵物流動統計」。
- Sample period: 民國 106 年 至 112 年（依資料內容可見，最早為 106 年 2 月，最晚為 112 年 2 月）
- Known issues:
    - end_dog_max_percent、end_cat_max_percent 可能超過 100%（顯示期末數量遠高於最大收容量）
    - 某些欄位（如 max_stay_dog_num、max_stay_cat_num）在部分縣市均為 0，推測無填報或資料缺失
    - 百分比欄位以字串格式呈現（含 %），需轉換為數值
    - 各縣市的欄位填報完整度差異大，例如部分縣市 fs_ms_num 長期為 0
- Definition for each variable:
    - ID: 資料唯一識別碼
    - rpt_year: 報表年份（民國年）
    - rpt_month: 報表月份
    - rpt_country_code: 縣市代碼
    - rpt_country: 縣市名稱
    - max_stay_dog_num: 犬隻最大容留量
    - max_stay_cat_num: 貓隻最大容留量
    - fs_gg_num: 期初犬隻數（fs = first stage or initial stock）
    - fs_ms_num: 期初貓隻數
    - fs_sum_num: 期初總收容量
    - in_gg_num: 當月犬隻移入數
    - in_ms_num: 當月貓隻移入數
    - in_lv_num: 移入類型—認領回或領養（需依官方定義確認）
    - in_bk_num: 移入類型—追回（如走失找回）
    - in_re_num: 移入類型—其他返還
    - in_lw_num: 移入類型—弱勢或拾獲來源
    - in_els_num: 移入類型—其他
    - in_tot_num: 移入總數
    - out_tback_num: 原飼主領回數
    - out_ad_ca_num: 犬隻認養數
    - out_ad_fa_num: 貓隻認養數
    - out_ad_cv_num: 其他類型認養數
    - out_hs_3_num: 移出類型—分類 3（需官方定義）
    - out_hs_5_num: 移出類型—分類 5
    - out_hs_7_num: 移出類型—分類 7
    - out_hs_ot_num: 移出類型—其他
    - out_sd_num: 犬貓死亡處理數
    - out_jd_num: 結紮後處置、行政結案或其他（依縣市差異）
    - out_rl_num: 野外釋放（release）
    - out_ec_num: 疾病相關移出
    - out_el_num: 其他移出
    - out_tot_num: 移出總數
    - fe_gg_num: 期末犬隻數
    - fe_ms_num: 期末貓隻數
    - fe_sum_num: 期末總數
    - end_dog_max_percent: 期末犬隻數 / 最大容留量（百分比字串）
    - end_cat_max_percent: 期末貓隻數 / 最大容留量（百分比字串）
