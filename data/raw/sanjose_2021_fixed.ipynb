# Data Description
- Background: 美國加州 San Jose 動物收容中心的動物個體資料，用於分析動物收容、進所原因、結局（Outcome）、品種、毛色、地點與健康狀態。
- date received: 20250928
- Path to data file: /data/animal_shelter/sanjose/sanjose_2021.csv
- Unit of observation: 每一筆資料代表「一隻動物的個體收容紀錄」。
- Sample period: 依資料內容顯示涵蓋 2020 年 8 月至 2021 年 3 月（不同個體略有差異）。
- Known issues:
    - SecondaryColor、DOB、OutcomeSubtype、IntakeReason 等欄位部分缺失。
    - Age 為字串格式（例如 “23 YEARS”），需轉換後才能進行量化分析。
    - Crossing（拾獲地點）為非結構化文字，具缺失情形。
    - Jurisdiction 欄位混雜不同城市（SAN JOSE、MILPITAS 等）。
    - 多個日期欄位需轉換為 datetime 格式以進行時間序列分析。

- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（可能缺失）
    - AnimalType: 動物種類（如 DOG、CAT）
    - PrimaryColor: 主要毛色
    - SecondaryColor: 次要毛色（部分缺失）
    - PrimaryBreed: 動物品種
    - Sex: 性別與結紮狀態（如 MALE、FEMALE、SPAYED、NEUTERED）
    - DOB: 出生日期（部分缺失）
    - Age: 動物年齡（字串格式）
    - IntakeDate: 進所日期
    - IntakeCondition: 進所時健康狀況（如 HEALTHY、UNHEALTHY、MED R）
    - IntakeType: 進所類型（STRAY、OWNER SUR 等）
    - IntakeSubtype: 進所子分類（OTC、FIELD 等）
    - IntakeReason: 進所原因（部分缺失，例如 EUTH OTHER）
    - OutcomeDate: 結局日期
    - OutcomeType: 結局類型（如 EUTH、RTO、DISPOSAL 等）
    - OutcomeSubtype: 結局子分類（部分缺失）
    - OutcomeCondition: 結局時狀態（如 DEAD、MANAGE、UNHEALTHY）
    - Crossing: 動物被發現或通報之地點（多為非結構化文字）
    - Jurisdiction: 管轄行政區（如 SAN JOSE、MILPITAS）
    - LastUpdate: 此筆資料最後更新日期
