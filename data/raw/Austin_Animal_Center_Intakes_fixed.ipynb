# Data Description
- Background: 美國德州 Austin Animal Center 的動物進所（Intake）資料，用於分析動物的拾獲地點、進所原因、進所時健康狀態、品種、毛色、年齡等屬性。常用於建構動物生命歷程分析、空間分布研究、以及與結局資料（Outcome）配對。
- date received: 20250928
- Path to data file: /data/animal_shelter/austin/Austin_Animal_Center_Intakes.csv
- Unit of observation: 每一筆資料代表「一隻動物的進所紀錄（Intake Record）」。
- Sample period: 依資料內容顯示涵蓋 2013 年至 2019 年（不同動物的進所月份略有差異）。
- Known issues:
    - Found Location（拾獲地點）為非結構化地址字串，可能需要地理解析（geocoding）才能做地圖分析。
    - Sex upon Intake 與 Age upon Intake 為字串格式，使用前需要解析。
    - Breed 及 Color 為文字描述（非分類編碼），可能需標準化。
    - 部分資料項缺失或不一致（如 Name、Found Location）。
    - 日期欄位 DateTime 為字串格式，需要轉換為 datetime。

- Definition for each variable:
    - Animal ID: 動物唯一識別碼
    - Name: 動物名稱（部分缺失）
    - DateTime: 動物進所的日期與時間
    - MonthYear: 字串格式的進所年月資訊（與 DateTime 重複資訊）
    - Found Location: 動物被拾獲或通報地點（需清洗以進行分析）
    - Intake Type: 進所方式（如 Stray、Owner Surrender 等）
    - Intake Condition: 進所時健康狀態（如 Normal、Sick）
    - Animal Type: 動物種類（Dog、Cat 等）
    - Sex upon Intake: 進所時的性別與結紮狀態（如 Neutered Male、Spayed Female）
    - Age upon Intake: 動物進所時的年齡（字串格式，如 "2 years"、"4 weeks"）
    - Breed: 動物品種（如 Beagle Mix、Domestic Shorthair Mix 等）
    - Color: 動物毛色（如 Tricolor、Calico）
