# Data Description
- Background: 美國加州 San Jose 動物收容中心的動物個體資料，用於分析動物之進所原因、品種、毛色、健康狀態、結局（Outcome），以及收容流程。
- date received: 20250928
- Path to data file: /data/animal_shelter/sanjose/sanjose_2223.csv
- Unit of observation: 每一筆資料代表「一隻收容動物的個體紀錄」，包含進所、結局、基本特徵與地理資訊。
- Sample period: 依資料內容顯示涵蓋 2022 年 7 月至 2023 年 6 月（不同個體紀錄略有差異）。
- Known issues:
    - 多數欄位存在缺失值（SecondaryColor、DOB、IntakeReason、OutcomeSubtype 等）。
    - Age 為字串格式（如 "19 YEARS"），需轉換後才能量化。
    - Crossing（拾獲地點）欄位非結構化且部分缺失。
    - Jurisdiction 部分資料為 NaN，需後續補齊或忽略。
    - 日期皆為字串格式，使用前需轉換為 datetime。
- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（部分缺失）
    - AnimalType: 動物種類（CAT、DOG 等）
    - PrimaryColor: 主要毛色
    - SecondaryColor: 次要毛色（部分缺失）
    - PrimaryBreed: 動物品種
    - Sex: 生理性別與結紮狀態（SPAYED、NEUTERED、MALE、FEMALE）
    - DOB: 出生日期（部分缺失）
    - Age: 年齡資訊（字串格式，如 "17 YEARS"）
    - IntakeDate: 動物進所日期
    - IntakeCondition: 進所時健康狀態（如 MED R、HEALTHY）
    - IntakeType: 進所類型（STRAY、DISPO REQ 等）
    - IntakeSubtype: 進所子分類（OTC、OTC OWNED 等）
    - IntakeReason: 進所原因（部分缺失）
    - OutcomeDate: 動物結局日期
    - OutcomeType: 結局類型（DISPOSAL、ADOPTION、RTO 等）
    - OutcomeSubtype: 結局子分類（部分缺失）
    - OutcomeCondition: 結局時狀態（如 DEAD、MED R）
    - Crossing: 拾獲或通報地點（部分缺失）
    - Jurisdiction: 所屬行政區域（如 SAN JOSE）
    - LastUpdate: 此筆資料最後更新日期
