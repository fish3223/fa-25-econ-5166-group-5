# Data Description
- Background: 美國加州 San Jose 動物收容中心之動物個體資料，用於分析動物的收容流動、進所動機、品種、毛色、健康狀態與後續結局（Outcome）。
- date received: 20250928
- Path to data file: /data/animal_shelter/sanjose/sanjose_1819.csv
- Unit of observation: 每一筆資料代表「一隻動物的個體收容紀錄」。
- Sample period: 依資料內容顯示涵蓋 2018 年 11 月至 2019 年 3 月（因個體紀錄不同而略有差異）。
- Known issues:
    - SecondaryColor、IntakeReason、OutcomeSubtype 等欄位部分缺失。
    - Age 為字串格式（如 "21 YEARS"），需清洗後方能用於量化分析。
    - Crossing（拾獲地點）欄位內容非結構化且缺失比例高。
    - Jurisdiction 欄位部分為 NaN，需補齊或忽略。
    - 日期欄位需轉換為 datetime 格式才能進行時間序列分析。

- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（部分為空值）
    - AnimalType: 動物種類（如 CAT、DOG）
    - PrimaryColor: 動物主要毛色
    - SecondaryColor: 動物次要毛色（部分缺失）
    - PrimaryBreed: 主要品種（如 DOMESTIC SH、SIAMESE 等）
    - Sex: 性別與結紮狀態（SPAYED、NEUTERED、MALE、FEMALE）
    - DOB: 出生日期（部分缺失）
    - Age: 動物年齡（字串格式，例如 “24 YEARS”）
    - IntakeDate: 進所日期
    - IntakeCondition: 進所狀態（如 HEALTHY、MANAGE 等）
    - IntakeType: 進所方式（STRAY、DISPO REQ 等）
    - IntakeSubtype: 進所子分類（OTC、FIELD 等）
    - IntakeReason: 進所原因（空值頻繁）
    - OutcomeDate: 動物結局日期
    - OutcomeType: 結局類型（RTO、DISPOSAL 等）
    - OutcomeSubtype: 結局子分類（多為空值）
    - OutcomeCondition: 結局時狀態（如 DEAD、MANAGE）
    - Crossing: 動物拾獲或通報地點（非結構性文字）
    - Jurisdiction: 管轄區域（如 SAN JOSE，部分為 NaN）
    - LastUpdate: 此筆資料最後更新時間
