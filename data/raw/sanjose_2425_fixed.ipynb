


# Data Description
- Background: 美國加州 San Jose 動物收容中心資料，用於分析動物之收容、進所原因、品種、健康狀態、結局（Outcome）等構成特徵。
- date received: 20250928
- Path to data file: /data/animal_shelter/sanjose/SanJose_2425.csv
- Unit of observation: 每一筆資料代表「一隻收容動物的個體紀錄」，包含其進所、結局、顏色、品種、性別、位置等特徵。
- Sample period: 依資料內容顯示，資料期間涵蓋 2024 年 8 月至 2024 年 12 月（不同動物紀錄日期略有差異）。
- Known issues:
    - DOB、SecondaryColor、IntakeReason、OutcomeSubtype 部分為缺失值。
    - Age 為字串格式（例如 "16 YEARS", "NO AGE"），需前處理後才能做數值化分析。
    - Crossing（拾獲地點）有缺失值或非結構化格式。
    - Date 欄位均為字串格式，需要轉換成 datetime 以便時間序列分析。
    - 品種（PrimaryBreed）類別數多，建模時可能需群組或重新編碼。

- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（可能缺失）
    - AnimalType: 動物種類（DOG、CAT 等）
    - PrimaryColor: 主要毛色
    - SecondaryColor: 次要毛色（部分缺失）
    - PrimaryBreed: 動物主要品種
    - Sex: 性別與結紮狀態（如 SPAYED、NEUTERED、MALE、FEMALE）
    - DOB: 動物出生日期（部分缺失）
    - Age: 動物年齡（字串格式）
    - IntakeDate: 動物進所日期
    - IntakeCondition: 進所時動物狀況（如 MEDICAL、HEALTHY）
    - IntakeType: 進所類型（STRAY、DISPO REQ 等）
    - IntakeSubtype: 進所子分類（OTC、OTC OWNED 等）
    - IntakeReason: 進所原因（部分缺失）
    - OutcomeDate: 動物結局日期
    - OutcomeType: 結局類型（RTO、ADOPTION、DISPOSAL 等）
    - OutcomeSubtype: 結局子分類（可能缺失）
    - OutcomeCondition: 結局時狀態（如 DEAD、MED R）
    - Crossing: 拾獲地點或通報地點
    - Jurisdiction: 所屬行政區域（如 SAN JOSE、MILPITAS）
    - LastUpdate: 該紀錄最後更新日期
