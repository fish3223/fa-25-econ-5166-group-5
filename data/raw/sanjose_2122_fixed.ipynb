# Data Description
- Background: 美國加州 San Jose 動物收容中心的動物個體資料，用於分析收容動物的進所、品種、毛色、健康狀態、通報地點，以及後續結局（Outcome）。
- date received: 20250928
- Path to data file: /data/animal_shelter/sanjose/sanjose_2122.csv
- Unit of observation: 每一筆資料代表「一隻動物的個體收容紀錄」。
- Sample period: 依資料內容顯示涵蓋 2021 年 7 月至 2021 年 12 月（不同個體略有差異）。
- Known issues:
    - SecondaryColor、DOB、IntakeReason、OutcomeSubtype 欄位大量缺失。
    - Age 為字串格式（如 "NO AGE"、"21 YEARS"），需轉換後才能做定量分析。
    - Crossing（拾獲地點）為非結構化文字，且部分缺失。
    - Jurisdiction 欄位部分缺失或為不同城市（SAN JOSE、CUPERTINO 等）。
    - 日期欄位需轉換為 datetime 格式才能進行時間序列分析。

- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（可能缺失）
    - AnimalType: 動物種類（如 DOG、CAT）
    - PrimaryColor: 主要毛色
    - SecondaryColor: 次要毛色（部分缺失）
    - PrimaryBreed: 動物主要品種
    - Sex: 性別與結紮狀態（如 SPAYED、NEUTERED、MALE、FEMALE）
    - DOB: 出生日期（部分缺失）
    - Age: 年齡資訊（字串格式，如 "NO AGE"）
    - IntakeDate: 進所日期
    - IntakeCondition: 進所時健康狀況（如 MED R、HEALTHY）
    - IntakeType: 進所方式（STRAY、DISPO REQ 等）
    - IntakeSubtype: 進所子分類（OTC、FIELD 等）
    - IntakeReason: 進所原因（部分缺失）
    - OutcomeDate: 結局日期
    - OutcomeType: 結局類型（如 DISPOSAL、RTO、ADOPTION）
    - OutcomeSubtype: 結局子分類（部分缺失）
    - OutcomeCondition: 結局時狀態（如 DEAD、MED R）
    - Crossing: 拾獲或通報地點
    - Jurisdiction: 管轄行政區（SAN JOSE、CUPERTINO 等）
    - LastUpdate: 資料最後更新時間
