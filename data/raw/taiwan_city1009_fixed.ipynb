

# Data Description
- Background: 台灣寵物資料。為分析台灣為各縣市對寵物之偏好。
- date received: 20251003
- Path to data file: /data/pet_management/taiwan/monthly_city_reports/taiwan_city1009.json
- Unit of observation: 每一筆資料代表「某一縣市在某一年月份的寵物流動統計」。
- Sample period: 民國 113 年 9 月 至 114 年 2 月
- Known issues:
    - 認養率（adopt_rate）可能超過 100%
    - 百分比欄位以字串格式（含 %）呈現，需轉換為數值
    - 部分縣市某些月份收容數為 0，可能無收容所或未通報
- Definition for each variable:
    - ID: 資料唯一識別碼
    - rpt_year: 報表年份（民國年）
    - rpt_month: 報表月份
    - rpt_country_code: 縣市代碼
    - rpt_country: 縣市名稱
    - accept_num: 當月收容動物數
    - adopt_num: 當月認養動物數
    - adopt_rate: 認養率（字串格式百分比）
    - adopt_total: 當月認養累計（資料中多為 0，可能為預留欄位）
    - end_num: 非認養之結案數（如移交或其他方式）
    - end_rate: 結案率（字串格式百分比）
    - dead_num: 當月死亡數
    - dead_rate: 死亡率（字串格式百分比）
