
# Data Description
- Background: 台灣各縣市收容所之動物個體資料，用於分析各地收容動物狀況（如品種、年齡、開放領養情形等）。
- date received: 20251009
- Path to data file: 原始資料在資料中心（如行資）之路徑。請填寫原始檔案之路徑。
- Unit of observation: 每一筆資料代表「一隻收容動物於收容所中的個體紀錄」。
- Sample period: 不同收容所資料更新時間不一，最早可追溯至民國 113 年，最新更新至 2025/10/09。
- Known issues:
    - 部分欄位（如 animal_foundplace、animal_title）可能為空字串。
    - animal_sterilization 與 animal_bacterin 可能為空值或標記不一致（T/N/F）。
    - album_file 可能為空字串，表示未提供照片。
    - animal_closeddate 多為 2999-12-31，為系統預設值並非實際關閉時間。
    - 部分資料可能因行政區代碼差異需額外比對 mapping 才能與縣市資料整併。

- Definition for each variable:
    - animal_id: 動物個體之唯一識別碼。
    - animal_subid: 收容所內部個體編號。
    - animal_area_pkid: 行政區域代碼（對應縣市）。
    - animal_shelter_pkid: 收容所代碼。
    - animal_place: 動物所在之收容所名稱。
    - animal_kind: 動物種類（例如：狗、貓）。
    - animal_Variety: 動物品種。
    - animal_sex: 生理性別（M/F/N）。
    - animal_bodytype: 體型（SMALL/MEDIUM/BIG）。
    - animal_colour: 毛色描述。
    - animal_age: 年齡階段（ADULT/CHILD）。
    - animal_sterilization: 是否結紮（T=已結紮，F=未結紮，N=未知）。
    - animal_bacterin: 是否施打疫苗（T/F）。
    - animal_foundplace: 通報拾獲地點。
    - animal_title: 系統標題（多為空值）。
    - animal_status: 動物狀態（如 OPEN 表示開放認養）。
    - animal_remark: 備註（健康狀況、受傷情形等）。
    - animal_caption: 備註補充欄位（多為空）。
    - animal_opendate: 開放認養日期。
    - animal_closeddate: 收容記錄關閉日期（多為系統預設）。
    - animal_update: 系統更新時間。
    - animal_createtime: 動物資料建立日期。
    - shelter_name: 收容所名稱。
    - album_file: 動物照片檔案網址。
    - album_update: 照片更新日期。
    - cDate: 系統同步日期。
    - shelter_address: 收容所地址。
    - shelter_tel: 收容所聯絡電話。
