# Data Description
- Background: 美國加州 San Jose 動物收容中心資料，用於分析動物收容、進所原因、品種、顏色、結局（Outcome）以及收容流程等特性。
- date received: 2025-9-28
- Path to data file: /data/animal_shelter/sanjose/sanjose_2526.csv
- Unit of observation: 每一筆資料代表「一隻動物的收容紀錄」，包含其進所、結局、基本特徵等資訊。
- Sample period: 依資料顯示，資料期間涵蓋 2025 年 7–8 月（可依需要延伸或以原始資料期間替換）。
- Known issues:
    - 部分欄位（如 SecondaryColor、IntakeReason、OutcomeSubtype、Crossing）存在缺失值
    - Age 為字串格式（如 "20 YEARS"），需前處理才能做年齡分析
    - IntakeSubtype、OutcomeSubtype 類別眾多，需標準化以利建模
    - DOB 與 Date 欄位為日期字串，需轉換 datetime 方可進行時間分析
- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（若未命名則可能為 NaN）
    - AnimalType: 動物種類（如 DOG, CAT）
    - PrimaryColor: 主要毛色
    - SecondaryColor: 次要毛色（可能為空值）
    - PrimaryBreed: 主要品種
    - Sex: 性別與是否結紮（如 SPAYED、NEUTERED、FEMALE、MALE）
    - DOB: 動物出生日期
    - Age: 動物年齡（字串格式）
    - IntakeDate: 動物進所日期
    - IntakeCondition: 進所時狀態（如 MED R、HEALTHY）
    - IntakeType: 進所類型（如 STRAY, DISPO REQ）
    - IntakeSubtype: 進所子分類（如 OTC、OTC OWNED）
    - IntakeReason: 進所原因（部分為空值）
    - OutcomeDate: 動物結局日期
    - OutcomeType: 結局類型（如 RTO, ADOPTION, DISPOSAL）
    - OutcomeSubtype: 結局子分類（如 RETURNED、FOSTER 等，部分空值）
    - OutcomeCondition: 結局時狀態（如 MED R, DEAD）
    - Crossing: 發現或拾獲地點資訊
    - Jurisdiction: 所屬管理區域（多為 SAN JOSE）
    - LastUpdate: 此筆資料最後更新日期
