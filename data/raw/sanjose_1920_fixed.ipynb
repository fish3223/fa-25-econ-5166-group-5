# Data Description
- Background: 美國加州 San Jose 動物收容中心之個體資料，用於分析動物進所原因、品種、毛色、年齡、健康狀態、拾獲地點及結局（Outcome）。
- date received: 20250928
- Path to data file: /data/animal_shelter/sanjose/sanjose_1920.csv
- Unit of observation: 每一筆資料代表「一隻動物的個體收容紀錄」。
- Sample period: 依資料內容顯示涵蓋 2019 年 10 月至 2020 年 05 月（因動物進出所時間不同，區間略有差異）。
- Known issues:
    - SecondaryColor、DOB、OutcomeSubtype、IntakeReason 等欄位部分缺失。
    - Age 為字串格式（如 "22 YEARS"），需標準化後才能量化處理。
    - Crossing（拾獲地點）為非結構化文字，格式不一致。
    - Jurisdiction 欄位皆為 SAN JOSE，但資料仍須留意跨城市之缺失或多樣性。
    - 日期欄位（IntakeDate、OutcomeDate、LastUpdate）需轉換為 datetime 格式。
- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（可能缺失）
    - AnimalType: 動物種類（DOG、CAT 等）
    - PrimaryColor: 主要毛色
    - SecondaryColor: 次要毛色（部分缺失）
    - PrimaryBreed: 動物品種（如 DOMESTIC SH、CHIHUAHUA SH 等）
    - Sex: 性別與結紮狀態（FEMALE、MALE、SPAYED、NEUTERED）
    - DOB: 出生日期（部分缺失）
    - Age: 動物年齡資訊（字串格式）
    - IntakeDate: 進所日期
    - IntakeCondition: 進所時健康狀態（如 HEALTHY、MED R 等）
    - IntakeType: 進所方式（STRAY、DISPO REQ 等）
    - IntakeSubtype: 進所子分類（OTC、FIELD 等）
    - IntakeReason: 進所原因（部分缺失，如 DOA）
    - OutcomeDate: 結局日期
    - OutcomeType: 結局方式（DISPOSAL、RTO 等）
    - OutcomeSubtype: 結局子分類（部分缺失）
    - OutcomeCondition: 結局狀態（如 DEAD、MANAGE 等）
    - Crossing: 動物發現或通報位置（非結構化文字）
    - Jurisdiction: 管轄區域（多為 SAN JOSE）
    - LastUpdate: 資料最後更新時間
