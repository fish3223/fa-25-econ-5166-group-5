# Data Description
- Background: 美國加州 San Jose 動物收容中心的動物個體資料，用於分析動物之進所、結局（Outcome）、品種、毛色、健康狀態與年齡特徵。
- date received: 20250928
- Path to data file: /data/animal_shelter/sanjose/sanjose_2324.csv
- Unit of observation: 每一筆資料代表「一隻收容動物的個體紀錄」。
- Sample period: 資料期間涵蓋 2023 年 7 月至 2024 年 3 月（不同動物紀錄略有差異）。
- Known issues:
    - 部分欄位缺失（如 SecondaryColor、DOB、IntakeReason、OutcomeSubtype）。
    - Age 為字串格式（如 "41 YEARS", "NO AGE"），需轉換後才能進行量化分析。
    - Crossing（拾獲地點）格式多樣，部分缺失。
    - Date 欄位（IntakeDate、OutcomeDate、LastUpdate）為字串格式，需轉換為 datetime。
    - Jurisdiction 欄位部分為 NaN，需與地理資料比對補齊。

- Definition for each variable:
    - AnimalID: 動物唯一識別碼
    - AnimalName: 動物名稱（可能缺失）
    - AnimalType: 動物種類（DOG、CAT 等）
    - PrimaryColor: 主要毛色
    - SecondaryColor: 次要毛色（可能缺失）
    - PrimaryBreed: 主要品種
    - Sex: 性別與結紮狀態（SPAYED、NEUTERED、MALE、FEMALE）
    - DOB: 動物出生日期（部分缺失）
    - Age: 動物年齡（字串格式）
    - IntakeDate: 動物進所日期
    - IntakeCondition: 進所狀態（如 MED R、HEALTHY）
    - IntakeType: 進所類型（STRAY、DISPO REQ、FOSTER 等）
    - IntakeSubtype: 進所子分類（OTC、OTC OWNED 等）
    - IntakeReason: 進所原因（部分缺失）
    - OutcomeDate: 動物結局日期
    - OutcomeType: 結局類型（如 DIED、RTO、ADOPTION、RESCUE）
    - OutcomeSubtype: 結局子分類（部分缺失）
    - OutcomeCondition: 結局時狀態（如 MED R、DEAD）
    - Crossing: 拾獲/通報地點（部分缺失）
    - Jurisdiction: 所屬行政區域（如 SAN JOSE）
    - LastUpdate: 此筆資料最後更新日期
