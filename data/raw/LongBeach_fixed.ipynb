# Data Description
- Background: 美國加州 Long Beach 動物收容中心的個體收容資料，用於分析動物進所原因、品種、毛色、地理位置（含經緯度）、健康狀況、死亡狀態與最終結局（Outcome）。
- date received: 20250928
- Path to data file: /data/animal_shelter/longbeach/LongBeach.csv
- Unit of observation: 每一筆資料代表「一隻動物的個體收容紀錄」。
- Sample period: 依資料內容顯示資料涵蓋 2017 年至 2023 年（不同個體紀錄略有差異）。
- Known issues:
    - Secondary Color、DOB、Outcome Subtype 等欄位部分缺失。
    - Sex 欄位格式不一致（如 Male、Female、Spayed、Neutered、Unknown）。
    - intake_is_dead、outcome_is_dead 與 was_outcome_alive 定義需依機構資料確認（布林值可能代表不同標準）。
    - Crossing（地點）資料來源不一致，部分缺失。
    - Jurisdiction 欄位可能非固定（需確認 Long Beach 官方定義）。
    - 日期欄位需轉換為 datetime 格式以進行時間序列分析。

- Definition for each variable:
    - Animal ID: 動物唯一識別碼
    - Animal Name: 動物名稱（部分缺失）
    - Animal Type: 動物種類（DOG、CAT、BIRD 等）
    - Primary Color: 主要毛色
    - Secondary Color: 次要毛色（部分缺失）
    - Sex: 動物性別與結紮狀態（如 Male、Female、Spayed、Neutered、Unknown）
    - DOB: 出生日期（部分缺失）
    - Intake Date: 進所日期
    - Intake Condition: 進所時動物狀態（如 NORMAL、UNDER AGE/WEIGHT、FERAL、INJURED SEVERE）
    - Intake Type: 進所方式（STRAY、WILDLIFE、OWNER SURRENDER 等）
    - Intake Subtype: 進所子分類（部分缺失）
    - Reason for Intake: 動物進所原因（如 OWNER SURRENDER REASON 等）
    - Outcome Date: 結局日期
    - Outcome Type: 結局類型（ADOPTION、RESCUE、EUTHANASIA、FOSTER TO ADOPT 等）
    - Outcome Subtype: 結局子分類（部分缺失）
    - Crossing: 拾獲或通報地點
    - Jurisdiction: 管轄區域（如 LONG BEACH 或其他）
    - latitude: 緯度（地理資訊）
    - longitude: 經度（地理資訊）
    - intake_is_dead: 進所時是否死亡（布林值）
    - outcome_is_dead: 結局是否死亡（布林值）
    - was_outcome_alive: 結局是否為存活狀態（布林值）
    - geopoint: 經緯度組合欄位（用於地圖繪製）
    - intake_duration: 動物在所天數
    - is_current_month: 是否屬於當月紀錄（0/1）
